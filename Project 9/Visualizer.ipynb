```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sbn
import numpy as np
```

```python
print("="*10,"Data analysis and visualization Program","="*10)
my_data = pd.read_csv(r"Product Sales Data.csv")
```

```python
while True:
    print("\nplease select an option :")
    print("1.Load dataset")
    print("2.Explore dataset")
    print("3.Perform dataset operations")
    print("4.Handle missing data")
    print("5.Generate Descriptive statistics")
    print("6.Data Visualization")
    print("7.save visualization")
    print("8.Exit\n")

    try:
        choice = int(input("Enter your choice: \n"))
    except ValueError:
        print("Please enter a valid menu number.")
        continue

    if choice == 1:
        file_name = input("Enter CSV file name (or press Enter to use Product Sales Data.csv): ")
        if file_name.strip() == "":
            file_name = "Product Sales Data.csv"
        try:
            my_data = pd.read_csv(file_name)
            print("Dataset loaded successfully.")
            
        except FileNotFoundError:
            print("File not found. Please check the file name and try again.")

    elif choice == 2:

        while True:

            print("\n1.Display first 5 rows of the dataset")
            print("2.Display last 5 rows of the dataset")
            print("3.Display column names of the dataset")
            print("4.Dispaly Data types of the dataset")
            print("5.Display basic information of the dataset")
            print("6.Exit\n")

            sub_choice = int(input("Enter your choice: \n"))

            if sub_choice == 1:
                print(my_data.head())

            elif sub_choice == 2:
                print(my_data.tail())

            elif sub_choice == 3:
                print(my_data.columns)

            elif sub_choice == 4:
                print(my_data.dtypes)
            
            elif sub_choice == 5:
                print(my_data.info())
            
            elif sub_choice == 6:
                break

            else:
                print("Invalid choice. Please try again.")

    elif choice == 3:

        while True:
            print("\n1. Create NumPy arrays, Indexing and Slicing")
            print("2. Mathematical Operations")
            print("3. Combine DataFrames")
            print("4. Split DataFrame")
            print("5. Search for Specific Values")
            print("6. Sort Data")
            print("7. Filter Data")
            print("8. Aggregating Functions")
            print("9. Exit\n")

            sub_choice = int(input("Enter your choice: "))

            if sub_choice == 1:

                print("\nAvailable numeric columns:")
                print(my_data.select_dtypes(include="number").columns)

                column = input("Enter numeric column name: ")

                if column in my_data.select_dtypes(include="number").columns:

                    arr = my_data[column].to_numpy()

                    print("\nNumPy Array:")
                    print(arr)

                    index = int(input("\nEnter index to access: "))
                    print("Element at index", index, ":", arr[index])

                    start = int(input("Enter slicing start index: "))
                    end = int(input("Enter slicing end index: "))

                    print("Sliced Array:")
                    print(arr[start:end])

                else:
                    print("Invalid numeric column.")

            elif sub_choice == 2:

                print("\nNumeric columns:")
                print(my_data.select_dtypes(include="number").columns)

                col1 = input("Enter first numeric column: ")
                col2 = input("Enter second numeric column: ")

                if col1 in my_data.columns and col2 in my_data.columns:

                    arr1 = my_data[col1].to_numpy()
                    arr2 = my_data[col2].to_numpy()

                    print("\nAddition:")
                    print(arr1 + arr2)

                    print("\nSubtraction:")
                    print(arr1 - arr2)

                    print("\nMultiplication:")
                    print(arr1 * arr2)

                    print("\nDivision:")
                    print(arr1 / arr2)

                else:
                    print("Invalid column name.")

            elif sub_choice == 3:

                print("\n1. Concatenate DataFrame")
                print("2. Merge DataFrame")
                print("3. Join DataFrame")

                combine_choice = int(input("Enter your choice: "))

                if combine_choice == 1:

                    df1 = my_data.iloc[:5]
                    df2 = my_data.iloc[5:10]

                    result = pd.concat([df1, df2])

                    print("\nDataFrames after concat:")
                    print(result)

                elif combine_choice == 2:

                    df1 = my_data[["Month", "Cream"]].copy()
                    df2 = my_data[["Month", "Detergent"]].copy()

                    result = pd.merge(df1,df2,on="Month")

                    print("\nDataFrames after merge:")
                    print(result)

                elif combine_choice == 3:

                    df1 = my_data[["Month", "Cream"]].copy()
                    df2 = my_data[["Month", "Detergent"]].copy()

                    df1 = df1.set_index("Month")
                    df2 = df2.set_index("Month")

                    result = df1.join(df2)

                    print("\nDataFrames after join:")
                    print(result)

                else:
                    print("Invalid choice.")

            elif sub_choice == 4:

                print("\n1. Split by Month")
                print("2. Split by Month")

                split_choice = int(input("Enter your choice: "))

                if split_choice == 1:

                    month = input("Enter month (e.g., Jan): ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]

                    print("\nData for", month)
                    print(result)

                elif split_choice == 2:

                    month = input("Enter month (e.g., Jan): ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]

                    print("\nData for month", month)
                    print(result)

                else:
                    print("Invalid choice.")

            elif sub_choice == 5:

                print("\n1. Search Month")
                print("2. Search Month")

                search_choice = int(input("Enter your choice: "))

                if search_choice == 1:

                    month = input("Enter month: ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]
                        
                    print(result)

                elif search_choice == 2:

                    month = input("Enter month: ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]

                    print(result)

                else:
                    print("Invalid choice.")

            elif sub_choice == 6:

                print("\nAvailable numeric columns:")
                print(my_data.select_dtypes(include="number").columns)

                column = input("Enter column to sort by: ")

                if column in my_data.columns:

                    order = input("Enter ascending or descending (a/d): ")

                    if order.lower() == "a":
                        result = my_data.sort_values(by=column,ascending=True)

                    else:
                        result = my_data.sort_values(by=column,ascending=False)

                    print(result)

                else:
                    print("Invalid column.")

            elif sub_choice == 7:

                print("\n1. Filter by Month")
                print("2. Filter by Month")

                filter_choice = int(input("Enter your choice: "))

                if filter_choice == 1:

                    month = input("Enter month: ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]

                    print(result)

                elif filter_choice == 2:

                    month = input("Enter month: ")

                    result = my_data[my_data["Month"].str.contains(month, case=False, na=False)]

                    print(result)

                else:
                    print("Invalid choice.")

            elif sub_choice == 8:

                print("\n1. Sum")
                print("2. Mean")
                print("3. Count")

                agg_choice = int(input("Enter your choice: "))

                print("\nNumeric columns:")
                print(my_data.select_dtypes(include="number").columns)

                column = input("Enter column name: ")

                if column in my_data.select_dtypes(include="number").columns:

                    if agg_choice == 1:
                        print("Sum:", my_data[column].sum())

                    elif agg_choice == 2:
                        print("Mean:", my_data[column].mean())

                    elif agg_choice == 3:
                        print("Count:", my_data[column].count())

                    else:
                        print("Invalid choice.")

                else:
                    print("Invalid numeric column.")

            elif sub_choice == 9:
                break

            else:
                print("Invalid choice. Please try again.")

    elif choice == 4:
        while True:
            print("\n1.Display rows with missing values")
            print("2.Fill missing values with mean")
            print("3.Drop rows with missing values")
            print("4.Replace missing values with a specific value")
            print("5.Exit\n")

            sub_choice = int(input("Enter your choice: \n"))

            if sub_choice == 1:
                missing_rows = my_data[my_data.isnull().any(axis=1)]
                print(missing_rows)

            elif sub_choice == 2:
                my_data.fillna(my_data.select_dtypes(include="number").mean(), inplace=True)
                print("Missing values filled with mean.")

            elif sub_choice == 3:
                my_data.dropna(inplace=True)
                print("Rows with missing values dropped.")

            elif sub_choice == 4:
                specific_value = input("Enter the value to replace missing values: ")
                my_data.fillna(specific_value, inplace=True)
                print(f"Missing values replaced with {specific_value}.")

            elif sub_choice == 5:
                break

            else:
                print("Invalid choice. Please try again.")


    elif choice == 5:
        while True:
    
            print("\n1. Display Statistical Summary")
            print("2. Standard Deviation")
            print("3. Variance")
            print("4. Percentile")
            print("5. Quantile")
            print("6. Exit\n")

            stat_choice = int(input("Enter your choice: "))

            if stat_choice == 1:

                print("\nDescriptive Statistics:")
                print(my_data.describe())

            elif stat_choice == 2:

                print("\nStandard Deviation:")
                print(my_data.select_dtypes(include="number").std())

            elif stat_choice == 3:

                print("\nVariance:")
                print(my_data.select_dtypes(include="number").var())
        
            elif stat_choice == 4:

                percentile = float( input("Enter percentile (0-100): "))

                print("\nPercentile:")
                print(my_data.select_dtypes(include="number").quantile(percentile / 100))

            elif stat_choice == 5:

                quantile = float(input("Enter quantile (0-1): "))

                print("\nQuantile:")
                print( my_data.select_dtypes(include="number") .quantile(quantile) )

            elif stat_choice == 6:
                break

            else:
                print("Invalid choice. Please try again.")

    elif choice == 6:

        while True:

            print("\n== Data Visualization ==")
            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Exit\n")

            sub_choice = int(input("Enter your choice: "))

            if sub_choice == 7:
                print("Exiting Data Visualization...")
                break

            elif sub_choice == 1:
                print("\n== Bar Plot ==")

                x_axis = input("Enter X-axis column name: ")
                y_axis = input("Enter Y-axis column name: ")

                if x_axis not in my_data.columns or y_axis not in my_data.columns:
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                print("Generating bar plot...")

                plt.bar(my_data[x_axis], my_data[y_axis])
                plt.xlabel(x_axis)
                plt.ylabel(y_axis)
                plt.title("Bar Plot")
                plt.xticks(rotation=45)
                plt.tight_layout()
                plt.show()

                print("Bar plot displayed successfully!")

            elif sub_choice == 2:
                print("\n== Line Plot ==")

                x_axis = input("Enter X-axis column name: ")
                y_axis = input("Enter Y-axis column name: ")

                if x_axis not in my_data.columns or y_axis not in my_data.columns:
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                print("Generating line plot...")

                plt.plot(my_data[x_axis], my_data[y_axis], marker='o')
                plt.xlabel(x_axis)
                plt.ylabel(y_axis)
                plt.title("Line Plot")
                plt.xticks(rotation=45)
                plt.tight_layout()
                plt.show()

                print("Line plot displayed successfully!")

            elif sub_choice == 3:
                print("\n== Scatter Plot ==")

                x_axis = input("Enter X-axis column name: ")
                y_axis = input("Enter Y-axis column name: ")

                if x_axis not in my_data.columns or y_axis not in my_data.columns:
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                if not pd.api.types.is_numeric_dtype(my_data[x_axis]) or not pd.api.types.is_numeric_dtype(my_data[y_axis]):
                    print("Scatter plot requires numeric columns. Please choose two numeric columns.")
                    continue
                print("Generating scatter plot...")

                plt.scatter(my_data[x_axis], my_data[y_axis])
                plt.xlabel(x_axis)
                plt.ylabel(y_axis)
                plt.title("Scatter Plot")
                plt.tight_layout()
                plt.show()

                print("Scatter plot displayed successfully!")

            elif sub_choice == 4:
                print("\n== Pie Chart ==")

                column = input("Enter column name: ")

                if column not in my_data.columns:
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                print("Generating pie chart...")

                my_data[column].value_counts().plot(kind='pie', autopct='%1.1f%%')

                plt.title("Pie Chart")
                plt.ylabel("")
                plt.tight_layout()
                plt.show()

                print("Pie chart displayed successfully!")

            elif sub_choice == 5:
                print("\n== Histogram ==")

                column = input("Enter column name: ")

                if column not in my_data.columns:
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                if not pd.api.types.is_numeric_dtype(my_data[column]):
                    print("Histogram requires a numeric column.")
                    continue
                print("Generating histogram...")

                plt.hist(my_data[column], bins=10, edgecolor='black')
                plt.xlabel(column)
                plt.ylabel("Frequency")
                plt.title("Histogram")
                plt.tight_layout()
                plt.show()

                print("Histogram displayed successfully!")

            elif sub_choice == 6:
                print("\n== Stack Plot ==")

                x_axis = input("Enter X-axis column name: ")
                y_axis = input("Enter Y-axis column names separated by comma: ")

                columns = [col.strip() for col in y_axis.split(",")]

                if x_axis not in my_data.columns or any(col not in my_data.columns for col in columns):
                    print("Invalid column name. Available columns:", list(my_data.columns))
                    continue
                if not pd.api.types.is_numeric_dtype(my_data[x_axis]) or any(not pd.api.types.is_numeric_dtype(my_data[col]) for col in columns):
                    print("Stack plot requires numeric columns for the X-axis and all Y-axis columns.")
                    continue
                print("Generating stack plot...")

                plt.stackplot(my_data[x_axis],*[my_data[col] for col in columns],labels=columns)

                plt.xlabel(x_axis)
                plt.ylabel("Values")
                plt.title("Stack Plot")
                plt.legend()
                plt.tight_layout()
                plt.show()

                print("Stack plot displayed successfully!")

            else:
                print("Invalid choice! Please try again.")

    elif choice == 7:

        print("\n== Save Visualization ==")

        file_name = input("Enter file name to save the plot (e.g., scatter_plot.png): ")

        if not file_name.endswith(".png"):
            file_name = file_name + ".png"

        plt.savefig(file_name, bbox_inches="tight")

        print(f"Visualization saved as {file_name} successfully!")

    elif choice == 8:
        print("Exiting the program...")
        break

    else:
        print("Invalid choice. Please try again.")


```
