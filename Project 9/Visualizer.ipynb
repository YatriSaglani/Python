```python
while True:
    print("Please select the type of data you want to analyze:")
    print("1.Load data")
    print("2.Explore data")
    print("3.Perform Dataframe operations")
    print("4.Handle missing values")
    print("5.Generate descriptive statistics")
    print("6.Data Visualizations")
    print("7.Save visualizations")
    print("8.Exit")

    choice = input("Enter your choice (1-8): ")

    if choice == '1':
        # Load data
        file_path = input("Enter the file path of the dataset: ")
        try:
            df = pd.read_csv(file_path)
            print("Data loaded successfully!")
        except Exception as e:
            print(f"Error loading data: {e}")

    elif choice == '2':
        # Explore data
        if 'df' in locals():
            print("Dataframe Info:")
            print(df.info())
            print("\nDataframe Head:")
            print(df.head())
        else:
            print("Please load the data first.")


    elif choice == '3':
        # Perform Dataframe operations
        if 'df' in locals():
            print("Select Dataframe operation:")
            print("1. Filter data")
            print("2. Sort data")
            print("3. Group data")
            operation_choice = input("Enter your choice (1-3): ")

            if operation_choice == '1':
                # Filter data
                column_name = input("Enter the column name to filter by: ")
                filter_value = input(f"Enter the value to filter {column_name} by: ")
                filtered_df = df[df[column_name] == filter_value]
                print("Filtered Dataframe:")
                print(filtered_df)

            elif operation_choice == '2':
                # Sort data
                column_name = input("Enter the column name to sort by: ")
                sorted_df = df.sort_values(by=column_name)
                print("Sorted Dataframe:")
                print(sorted_df)

            elif operation_choice == '3':
                # Group data
                column_name = input("Enter the column name to group by: ")
                grouped_df = df.groupby(column_name).size()
                print("Grouped Dataframe:")
                print(grouped_df)

            else:
                print("Invalid choice. Please select a valid operation.")

        else:
            print("Please load the data first.")

    elif choice == '4':
        # Handle missing values
        if 'df' in locals():
            print("Select missing value handling method:")
            print("1. Drop missing values")
            print("2. Fill missing values")
            missing_choice = input("Enter your choice (1-2): ")

            if missing_choice == '1':
                df = df.dropna()
                print("Missing values dropped.")
            elif missing_choice == '2':
                fill_value = input("Enter the value to fill missing values with: ")
                df = df.fillna(fill_value)
                print("Missing values filled.")
            else:
                print("Invalid choice. Please select a valid method.")

        else:
            print("Please load the data first.")
    
    elif choice == '5':
        # Generate descriptive statistics
        if 'df' in locals():
            print("Descriptive Statistics:")
            print(df.describe())
        else:
            print("Please load the data first.")

    elif choice == '6':
        # Data Visualizations
        if 'df' in locals():
            print("Select visualization type:")
            print("1. Histogram")
            print("2. Scatter plot")
            print("3. Box plot")
            viz_choice = input("Enter your choice (1-3): ")

            if viz_choice == '1':
                column_name = input("Enter the column name for histogram: ")
                df[column_name].hist()
                plt.title(f"Histogram of {column_name}")
                plt.xlabel(column_name)
                plt.ylabel("Frequency")
                plt.show()

            elif viz_choice == '2':
                x_column = input("Enter the column name for x-axis: ")
                y_column = input("Enter the column name for y-axis: ")
                df.plot.scatter(x=x_column, y=y_column)
                plt.title(f"Scatter Plot of {y_column} vs {x_column}")
                plt.show()

            elif viz_choice == '3':
                column_name = input("Enter the column name for box plot: ")
                df.boxplot(column=column_name)
                plt.title(f"Box Plot of {column_name}")
                plt.show()

            else:
                print("Invalid choice. Please select a valid visualization type.")

        else:
            print("Please load the data first.")

    elif choice == '7':
        # Save visualizations
        if 'df' in locals():
            save_choice = input("Do you want to save the last visualization? (yes/no): ")
            if save_choice.lower() == 'yes':
                file_name = input("Enter the file name to save the visualization (with extension .png or .jpg): ")
                plt.savefig(file_name)
                print(f"Visualization saved as {file_name}.")
            else:
                print("Visualization not saved.")
        else:
            print("Please load the data first.")

    elif choice == '8':
        print("Exiting the program. Goodbye!")
        break
```
