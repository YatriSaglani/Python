```python
import numpy as np

print("Welcome to NumPy Analyzer")


def arraycreation():

    print("\nArray Creation:")
    print("1. 1D Array")
    print("2. 2D Array")
    print("3. 3D Array")

    choice = int(input("Enter your choice: "))

    # 1D Array
    if choice == 1:

        n = int(input("\nEnter the number of elements: "))

        elements = list(map(int, input(
            f"Enter {n} elements separated by space: "
        ).split()))

        arr = np.array(elements)

    # 2D Array
    elif choice == 2:

        rows = int(input("\nEnter the number of rows: "))
        cols = int(input("Enter the number of columns: "))

        elements = list(map(int, input(
            f"Enter {rows * cols} elements separated by space: "
        ).split()))

        arr = np.array(elements).reshape(rows, cols)

    # 3D Array
    elif choice == 3:

        layers = int(input("\nEnter the number of layers: "))
        rows = int(input("Enter the number of rows: "))
        cols = int(input("Enter the number of columns: "))

        elements = list(map(int, input(
            f"Enter {layers * rows * cols} elements separated by space: "
        ).split()))

        arr = np.array(elements).reshape(layers, rows, cols)

    else:

        print("Invalid choice!")
        return None

    print("\nArray created successfully:")
    print(arr)

    # Operations
    while True:

        print("\nChoose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        operation = int(input("Enter your choice: "))

        # Indexing
        if operation == 1:

            index = input("Enter index: ")

            index = tuple(map(int, index.split()))

            print("\nElement at index:", arr[index])

        # Slicing
        elif operation == 2:

            if arr.ndim == 1:

                start, end = map(
                    int,
                    input("Enter the range (start:end): ").split(":")
                )

                print("\nSliced Array:")
                print(arr[start:end])

            elif arr.ndim == 2:

                row_start, row_end = map(
                    int,
                    input("Enter the row range (start:end): ").split(":")
                )

                col_start, col_end = map(
                    int,
                    input("Enter the column range (start:end): ").split(":")
                )

                print("\nSliced Array:")
                print(arr[row_start:row_end, col_start:col_end])

            elif arr.ndim == 3:

                print("For 3D array, enter slicing for each dimension.")

                d_start, d_end = map(
                    int,
                    input("Enter depth range (start:end): ").split(":")
                )

                r_start, r_end = map(
                    int,
                    input("Enter row range (start:end): ").split(":")
                )

                c_start, c_end = map(
                    int,
                    input("Enter column range (start:end): ").split(":")
                )

                print("\nSliced Array:")
                print(
                    arr[
                        d_start:d_end,
                        r_start:r_end,
                        c_start:c_end
                    ]
                )

        # Go Back
        elif operation == 3:

            break

        else:

            print("Invalid choice!")

    return arr


def mathematicaloperations(arr):

    print("\nMathematical Operations:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter your choice: "))

    size = arr.size

    elements = list(map(int, input(
        f"\nEnter the same-size array elements "
        f"({size} elements separated by space): "
    ).split()))

    second_array = np.array(elements).reshape(arr.shape)

    print("\nOriginal Array:")
    print(arr)

    print("\nSecond Array:")
    print(second_array)

    if choice == 1:

        result = arr + second_array

        print("\nResult of Addition:")
        print(result)

    elif choice == 2:

        result = arr - second_array

        print("\nResult of Subtraction:")
        print(result)

    elif choice == 3:

        result = arr * second_array

        print("\nResult of Multiplication:")
        print(result)

    elif choice == 4:

        if np.any(second_array == 0):

            print("Division by zero is not allowed.")

        else:

            result = arr / second_array

            print("\nResult of Division:")
            print(result)

    else:

        print("Invalid choice!")


def combine_split_arrays(arr):

    print("\nCombine or Split Arrays:")
    print("1. Combine Arrays")
    print("2. Split Array")

    choice = int(input("Enter your choice: "))

    # Combine Arrays
    if choice == 1:

        size = arr.size

        elements = list(map(int, input(
            f"\nEnter the elements of another array to combine "
            f"({size} elements separated by space): "
        ).split()))

        second_array = np.array(elements).reshape(arr.shape)

        print("\nOriginal Array:")
        print(arr)

        print("\nSecond Array:")
        print(second_array)

        combined_array = np.vstack((arr, second_array))

        print("\nCombined Array (Vertical Stack):")
        print(combined_array)

    # Split Array
    elif choice == 2:

        parts = int(input(
            "\nEnter the number of parts to split the array: "
        ))

        split_array = np.array_split(arr, parts)

        print("\nOriginal Array:")
        print(arr)

        print("\nSplit Arrays:")

        for i, part in enumerate(split_array, 1):

            print(f"\nArray {i}:")
            print(part)

    else:

        print("Invalid choice!")


def search_sort_filter(arr):

    print("\nSearch, Sort, and Filter:")
    print("1. Search a value")
    print("2. Sort the array")
    print("3. Filter values")

    choice = int(input("Enter your choice: "))

    print("\nOriginal Array:")
    print(arr)

    if choice == 1:

        value = int(input("\nEnter value to search: "))

        result = np.where(arr == value)

        if len(result[0]) > 0:

            print("Value found at index:", result)

        else:

            print("Value not found.")

    elif choice == 2:

        if arr.ndim == 1:

            sorted_array = np.sort(arr)

        else:

            sorted_array = np.sort(arr, axis=-1)

        print("\nSorted Array:")
        print(sorted_array)

        print("(Sorting applied row-wise.)")

    elif choice == 3:

        value = int(input("\nEnter value to filter: "))

        filtered_array = arr[arr > value]

        print("\nFiltered Array:")
        print(filtered_array)

    else:

        print("Invalid choice!")


def aggregates_statistics(arr):

    print("\nAggregates and Statistics:")
    print("1. Sum")
    print("2. Mean")
    print("3. Median")
    print("4. Standard Deviation")
    print("5. Variance")

    choice = int(input("Enter your choice: "))

    print("\nOriginal Array:")
    print(arr)

    if choice == 1:

        print("\nSum of Array:", np.sum(arr))

    elif choice == 2:

        print("\nMean of Array:", np.mean(arr))

    elif choice == 3:

        print("\nMedian of Array:", np.median(arr))

    elif choice == 4:

        print("\nStandard Deviation of Array:", np.std(arr))

    elif choice == 5:

        print("\nVariance of Array:", np.var(arr))

    else:

        print("Invalid choice!")


# Main Program

arr = None

while True:

    print("\n--------------------------------")
    print("       NUMPY ANALYZER")
    print("--------------------------------")

    print("1. Create a numpy array")
    print("2. Perform Mathematical functions")
    print("3. Combine or split arrays")
    print("4. Search, sort or filter arrays")
    print("5. Compute aggregates and statistics")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        arr = arraycreation()

    elif choice == 2:

        if arr is None:
            print("\nPlease create an array first.")

        else:
            mathematicaloperations(arr)

    elif choice == 3:

        if arr is None:
            print("\nPlease create an array first.")

        else:
            combine_split_arrays(arr)

    elif choice == 4:

        if arr is None:
            print("\nPlease create an array first.")

        else:
            search_sort_filter(arr)

    elif choice == 5:

        if arr is None:
            print("\nPlease create an array first.")

        else:
            aggregates_statistics(arr)

    elif choice == 6:

        print("\nThank you for using NumPy Analyzer!")
        break

    else:

        print("\nWrong choice!")
```
